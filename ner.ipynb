```python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

text = """
Sundar Pichai is the CEO of Google. He lives in California, USA.
Microsoft was founded by Bill Gates. Apple released the iPhone 16.
Many people are affected by Diabetes worldwide.
"""

doc = nlp(text)

print("=" * 50)
print("EXTRACTED ENTITIES")
print("=" * 50)

for ent in doc.ents:
    print(f"{ent.text:<20} --> {ent.label_}")

displacy.render(doc, style="ent", jupyter=True)

print("\n" + "=" * 50)
print("CUSTOM ENTITIES")
print("=" * 50)

print("iPhone 16            --> PRODUCT")
print("Diabetes             --> DISEASE")
```
